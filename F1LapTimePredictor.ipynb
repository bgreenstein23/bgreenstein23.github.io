```python

import pandas as pd


# 2. Load a session (Year, Event, Session type)
session = fastf1.get_session(2024, 'Silverstone', 'R')
session.load()

# 3. Get all laps
laps = session.laps

# 4. Create your 'lap_seconds' and 'next_lap_seconds' target
laps['lap_seconds'] = laps['LapTime'].dt.total_seconds()

# Sort by driver and lap to ensure the shift is correct
laps = laps.sort_values(by=['Driver', 'LapNumber'])
laps['next_lap_seconds'] = laps.groupby('Driver')['lap_seconds'].shift(-1)

# 5. Filter for clean data (TrackStatus 1 means normal racing, no Safety Car)
laps_clean = laps[laps['TrackStatus'] == '1'].copy()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Plotting Tyre Age vs Lap Time for the top 3 teams
top_teams = ['Red Bull Racing', 'McLaren', 'Mercedes']
eda_data = laps_clean[laps_clean['Team'].isin(top_teams)]

plt.figure(figsize=(10, 6))
sns.lineplot(data=eda_data, x='TyreLife', y='lap_seconds', hue='Compound')
plt.title('EDA: How Tire Age impacts Lap Time')
plt.show()
```

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


year = 2024
schedule = fastf1.get_event_schedule(year)

all_race_laps = []

# Load data for the first 5 races of the season
for i, event in schedule.iloc[2:7].iterrows():
    print(f"Loading {event['EventName']}...")
    try:
        session = fastf1.get_session(year, event['EventName'], 'R')
        session.load(laps=True, telemetry=False)

        # Keep only relevant columns and filter for green-flag laps (TrackStatus '1')
        laps = session.laps.pick_track_status('1')[['Driver', 'LapTime', 'Compound', 'LapNumber']]
        laps['RaceName'] = event['EventName']
        all_race_laps.append(laps)
    except Exception as e:
        print(f"Error loading {event['EventName']}: {e}")

# Combine into one DataFrame and convert lap times to seconds
full_data = pd.concat(all_race_laps)
full_data['lap_seconds'] = full_data['LapTime'].dt.total_seconds()
full_data = full_data.dropna(subset=['lap_seconds'])
```

```python
# Create a multi-plot grid (FacetGrid)
g = sns.FacetGrid(full_data, col="RaceName", col_wrap=3, height=4, sharex=False)
g.map(sns.histplot, "lap_seconds", bins=50, kde=True, color="red")

g.set_axis_labels("Lap Time (Seconds)", "Frequency")
g.set_titles("{col_name}")
plt.subplots_adjust(top=0.9)
g.fig.suptitle('Lap Time Distribution per Race (2024 Season)', fontsize=16)

plt.show()
```
